### Create a custom decorator for function result caching: if a specific function has already been executed with certain parameters, do not execute the function again; instead, return the result from the cache. Test it on 3-4 functions with varying numbers and types of parameters.

```python
from functools import wraps
def cache_result(func):

    cache = {}

    @wraps(func)
    def wrapper(*args, **kwargs):

        key = (args, tuple(sorted(kwargs.items())))

        if key in cache:
            print("Result returned from cache.")
            return cache[key]

        print("Function is executing...")

        result = func(*args, **kwargs)

        cache[key] = result

        return result

    return wrapper

@cache_result
def add(a, b):
    print(f"Calculating {a} + {b}")
    return a + b

print("=" * 60)
print("FUNCTION 1: ADD")
print("=" * 60)

print("\nFirst call:")
print("Result:", add(5, 10))

print("\nSecond call with the same parameters:")
print("Result:", add(5, 10))

print("\nThird call with different parameters:")
print("Result:", add(7, 10))

@cache_result
def greet(name):
    print(f"Creating greeting for {name}")
    return f"Hello, {name}!"

print("\n" + "=" * 60)
print("FUNCTION 2: GREET")
print("=" * 60)

print("\nFirst call:")
print(greet("Sofia"))

print("\nSecond call with the same parameter:")
print(greet("Sofia"))

print("\nThird call with a different parameter:")
print(greet("Anna"))

@cache_result
def calculate_average(a, b, c):
    print(f"Calculating average of {a}, {b}, {c}")
    return (a + b + c) / 3

print("\n" + "=" * 60)
print("FUNCTION 3: CALCULATE AVERAGE")
print("=" * 60)

print("\nFirst call:")
print("Average:", calculate_average(10, 20, 30))

print("\nSecond call with the same parameters:")
print("Average:", calculate_average(10, 20, 30))

print("\nThird call with different parameters:")
print("Average:", calculate_average(5, 15, 25))

@cache_result
def create_profile(name, age, student):
    print("Creating profile...")

    return {
        "name": name,
        "age": age,
        "student": student
    }

print("\n" + "=" * 60)
print("FUNCTION 4: CREATE PROFILE")
print("=" * 60)

print("\nFirst call:")
print(create_profile("Sofia", 19, True))

print("\nSecond call with the same parameters:")
print(create_profile("Sofia", 19, True))

print("\nThird call with different parameters:")
print(create_profile("Anna", 20, False))

```

## Conclusion

During this laboratory work, a custom decorator for caching function results was implemented in Python.

The decorator stores the results of previously executed functions in a cache using their parameters as keys. When a function is called again with the same parameters, the cached result is returned instead of executing the function again. If the parameters are different, the function is executed and the new result is stored in the cache.

The decorator was tested with four different functions using different numbers and types of parameters, including integers, strings, floating-point numbers, and Boolean values. Keyword arguments were also tested.

The results of the experiments demonstrated that the caching mechanism works correctly and prevents unnecessary repeated function executions. Thus, the main objective of the laboratory work was successfully achieved.
